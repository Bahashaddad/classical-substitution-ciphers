# Traditional substitution exercise
Educational only. This notebook uses synthetic text and the shared, tested implementation. The mapping and pair-swapping operations are not modern encryption.

```python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd() / "ciphers.py").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from ciphers import transform, analyze_text
plaintext = (root / "examples" / "plaintext.txt").read_text(encoding="utf-8")
ciphertext = transform(plaintext, modified=False)
recovered = transform(ciphertext, decrypt=True, modified=False)
assert recovered == plaintext.lower()
print(ciphertext)
print(analyze_text(ciphertext))
```
